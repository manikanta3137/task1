```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("wc_2026_fixtures.csv")

print(df.head())
print(df.info())

```

```
df.isnull().sum()
```

```
df["group"] = df["group"].fillna("Knockout")
```

```
df["date"] = pd.to_datetime(df["date"])
```

```
df.drop_duplicates(inplace=True)
```

```
df["stage"].value_counts()
```

```
df["country"].value_counts()
```

```
df["city"].value_counts()
```

```
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="stage",
    order=df["stage"].value_counts().index
)

plt.xticks(rotation=45)
plt.title("Matches by Tournament Stage")
plt.show()
```

```
plt.figure(figsize=(8,5))

df["country"].value_counts().plot(
    kind="bar"
)

plt.title("Matches Hosted by Country")
plt.ylabel("Matches")
plt.show()
```

```
rankings = pd.concat([
    df["team1_fifa_rank"],
    df["team2_fifa_rank"]
])

plt.figure(figsize=(8,5))

sns.histplot(rankings, bins=15)

plt.title("Distribution of FIFA Rankings")
plt.show()
```

```
confed = pd.concat([
    df["team1_confederation"],
    df["team2_confederation"]
])

confed.value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Confederation Representation")
plt.show()
```
